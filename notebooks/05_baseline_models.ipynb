## 1. Introduction

### 1.1 Purpose of This Notebook
In this notebook, we establish the baseline XGBoost prediction model. The baseline model is trained using only typical technical indicators (RSI, MACD, Bollinger Bands, ATR, realized vol). This model represents standard practice in index forecasting literature and will serve as the benchmark for evaluation.

## 2. Environment Setup

```python
import sys
from pathlib import Path
import pandas as pd
import joblib

project_dir = Path("e:/BN_Structure_Reforms/banknifty-expiry-volatility-ml")
if str(project_dir) not in sys.path:
    sys.path.append(str(project_dir))

from src import config
from src.models import build_xgb_regressor, train_model
from src.evaluation import calculate_metrics
```

## 3. Data Preparation

```python
train = pd.read_parquet(config.PROCESSED_DATA_DIR / "train.parquet")
val = pd.read_parquet(config.PROCESSED_DATA_DIR / "val.parquet")
test = pd.read_parquet(config.PROCESSED_DATA_DIR / "test.parquet")

# Define technical features only
baseline_features = ['RSI', 'MACD', 'MACD_Signal', 'MACD_Hist', 'BB_Mid', 'BB_Upper', 'BB_Lower', 'BB_PctB', 'BB_Width', 'ATR', 'Rolling_Vol', 'Return']

X_train, y_train = train[baseline_features], train['Target']
X_val, y_val = val[baseline_features], val['Target']
X_test, y_test = test[baseline_features], test['Target']
```

## 4. Baseline Model Training

```python
baseline_model = build_xgb_regressor()
train_model(baseline_model, X_train, y_train, X_val, y_val)
print("[OK] Baseline model trained successfully.")
```

## 5. Baseline Model Performance Evaluation

```python
test_preds = baseline_model.predict(X_test)
metrics = calculate_metrics(y_test, test_preds)
print("Baseline model test set metrics:", metrics)
joblib.dump(baseline_model, config.MODELS_DIR / "baseline_xgb.joblib")
```
